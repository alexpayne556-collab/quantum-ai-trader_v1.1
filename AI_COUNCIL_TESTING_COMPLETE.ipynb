# 🔬 PATTERN DISCOVERY ENGINE

**OLD WAY:** Test Perplexity's institutional detector (we already know what we're looking for)

**NEW WAY:** Let the DATA tell us what predicts winners

**Mission:** Find the UNKNOWN patterns that nobody else sees

**Method:**
1. Load 196 tickers
2. For each ticker, calculate 50+ features (price patterns, volume, volatility, sector, etc.)
3. Find which features ACTUALLY predict next-day winners
4. Build a "winner profile" from the data
5. Discover correlations we never thought to look for

**This is how we find the edge nobody else has.**

```python
class PatternDiscoveryEngine:
    """
    INNOVATIVE: Don't test our ideas - let the DATA show us what works
    
    This engine:
    1. Calculates 50+ features for every ticker/day
    2. Tracks which features predict next-day winners
    3. Finds UNKNOWN correlations
    4. Builds a data-driven "winner profile"
    """
    
    def __init__(self):
        self.feature_importance = {}
        self.discovered_patterns = []
    
    def extract_features(self, ticker, price_data, volume_data, days_back=60):
        """Extract 50+ features to find what ACTUALLY predicts winners"""
        
        df = pd.DataFrame({
            'price': price_data,
            'volume': volume_data
        })
        
        if len(df) < 20:
            return None
        
        features = {}
        
        # === PRICE PATTERNS ===
        # 1-5: Price position vs moving averages
        for ma_period in [5, 10, 20, 50, 200]:
            ma = df['price'].rolling(ma_period).mean().iloc[-1]
            if not pd.isna(ma):
                features[f'price_vs_ma{ma_period}'] = (df['price'].iloc[-1] / ma) - 1
        
        # 6-10: Recent price changes
        for days in [1, 2, 3, 5, 10]:
            if len(df) > days:
                features[f'price_change_{days}d'] = (df['price'].iloc[-1] / df['price'].iloc[-days-1]) - 1
        
        # 11: Volatility (20-day std)
        features['volatility_20d'] = df['price'].pct_change().tail(20).std()
        
        # 12: Price range (high-low amplitude)
        features['range_amplitude'] = (df['price'].tail(20).max() / df['price'].tail(20).min()) - 1
        
        # 13-15: Consecutive up/down days
        changes = df['price'].pct_change()
        features['consecutive_up'] = (changes.tail(10) > 0).sum()
        features['consecutive_down'] = (changes.tail(10) < 0).sum()
        features['streak_length'] = self._get_current_streak(changes)
        
        # === VOLUME PATTERNS ===
        # 16-20: Volume vs moving averages
        for ma_period in [5, 10, 20, 50]:
            vol_ma = df['volume'].rolling(ma_period).mean().iloc[-1]
            if not pd.isna(vol_ma) and vol_ma > 0:
                features[f'vol_vs_ma{ma_period}'] = (df['volume'].iloc[-1] / vol_ma)
        
        # 21: Volume trend (increasing/decreasing)
        vol_ma5 = df['volume'].rolling(5).mean()
        vol_ma20 = df['volume'].rolling(20).mean()
        if len(vol_ma5) > 0 and len(vol_ma20) > 0:
            features['vol_trend'] = (vol_ma5.iloc[-1] / vol_ma20.iloc[-1]) if vol_ma20.iloc[-1] > 0 else 1
        
        # 22: Volume spikes (days with >2x avg volume)
        vol_ma = df['volume'].rolling(20).mean()
        features['volume_spikes'] = ((df['volume'] / vol_ma) > 2).tail(10).sum()
        
        # 23: Volume consistency (lower std = more consistent)
        features['vol_consistency'] = 1 / (df['volume'].tail(20).std() / df['volume'].tail(20).mean() + 0.01)
        
        # === MOMENTUM INDICATORS ===
        # 24: RSI-like (overbought/oversold)
        gains = changes[changes > 0].tail(14).sum()
        losses = abs(changes[changes < 0].tail(14).sum())
        features['rsi'] = gains / (gains + losses + 0.001) if (gains + losses) > 0 else 0.5
        
        # 25-26: Price vs VWAP
        vwap = (df['price'] * df['volume']).cumsum() / df['volume'].cumsum()
        if len(vwap) > 0:
            features['price_vs_vwap'] = (df['price'].iloc[-1] / vwap.iloc[-1]) - 1
            features['days_below_vwap'] = (df['price'].tail(10) < vwap.tail(10)).sum()
        
        # 27: Acceleration (is momentum increasing?)
        ret_5d = df['price'].pct_change(5).iloc[-1]
        ret_10d = df['price'].pct_change(10).iloc[-1]
        features['acceleration'] = ret_5d - (ret_10d / 2) if not pd.isna(ret_5d) and not pd.isna(ret_10d) else 0
        
        # === TIME-BASED PATTERNS ===
        # 28: Day of week effect (Mon=0, Fri=4)
        features['day_of_week'] = df.index[-1].dayofweek if hasattr(df.index[-1], 'dayofweek') else 2
        
        # 29: Time since last big move (>5%)
        big_moves = abs(changes) > 0.05
        features['days_since_big_move'] = (len(changes) - big_moves[::-1].idxmax() - 1) if big_moves.any() else 99
        
        # === DISCOVERED PATTERNS (we'll add these as we find them) ===
        # 30: Gap behavior (open vs prev close)
        if len(df) > 1:
            # Assuming we have open prices, otherwise approximate
            features['gap_tendency'] = changes.tail(10).mean()
        
        # 31-35: Custom ratio combinations (experimental)
        features['vol_price_ratio'] = (features['vol_vs_ma20'] * features.get('price_change_1d', 0)) if 'vol_vs_ma20' in features else 0
        features['momentum_vol_combo'] = features['rsi'] * features['vol_trend'] if 'vol_trend' in features else 0
        features['streak_vol_combo'] = features['streak_length'] * features['vol_vs_ma5'] if 'vol_vs_ma5' in features else 0
        
        # 36-37: Sub-sector (fine granularity) + Broad sector
        sub_sector = get_sub_sector(ticker)
        broad_sector = get_broad_sector(ticker)
        features['sub_sector'] = hash(sub_sector) % 100  # Fine granularity
        features['broad_sector'] = hash(broad_sector) % 20  # Broad grouping
        
        # 37-40: Statistical moments
        features['skewness'] = changes.tail(20).skew() if len(changes) >= 20 else 0
        features['kurtosis'] = changes.tail(20).kurt() if len(changes) >= 20 else 0
        features['percentile_90'] = df['price'].tail(20).quantile(0.9) / df['price'].iloc[-1] if len(df) >= 20 else 1
        features['percentile_10'] = df['price'].tail(20).quantile(0.1) / df['price'].iloc[-1] if len(df) >= 20 else 1
        
        return features
    
    def _get_current_streak(self, changes):
        """How many consecutive up or down days?"""
        if len(changes) < 2:
            return 0
        streak = 1
        direction = 1 if changes.iloc[-1] > 0 else -1
        for i in range(len(changes)-2, -1, -1):
            if (changes.iloc[i] > 0 and direction > 0) or (changes.iloc[i] < 0 and direction < 0):
                streak += 1
            else:
                break
        return streak * direction  # Positive for up-streak, negative for down-streak
    
    def discover_winning_patterns(self, tickers, days_back=60):
        """
        MAGIC HAPPENS HERE:
        1. Calculate features for all tickers
        2. Track which features predict next-day winners
        3. Find correlations we never thought to look for
        """
        print("🔬 PATTERN DISCOVERY ENGINE")
        print("="*70)
        print(f"Analyzing {len(tickers)} tickers to find UNKNOWN patterns...")
        print("This will take a few minutes - we're calculating 40+ features per ticker\n")
        
        all_data = []
        processed = 0
        
        for ticker in tickers:
            data = get_data(ticker, days_back=days_back)
            if data is None or len(data) < 30:
                continue
            
            processed += 1
            if processed % 20 == 0:
                print(f"   Processed {processed}/{len(tickers)}...")
            
            # Calculate features for each day (we'll predict next day)
            for i in range(20, len(data)-1):
                features = self.extract_features(
                    ticker,
                    data['Close'].iloc[:i+1].values,
                    data['Volume'].iloc[:i+1].values,
                    days_back=i+1
                )
                
                if features is None:
                    continue
                
                # NEXT DAY OUTCOME (what we're trying to predict)
                next_day_return = (data['Close'].iloc[i+1] / data['Close'].iloc[i]) - 1
                
                all_data.append({
                    'ticker': ticker,
                    'date': data.index[i],
                    'next_day_return': next_day_return,
                    'is_winner': next_day_return > 0.02,  # >2% gain = winner
                    **features
                })
        
        df = pd.DataFrame(all_data)
        
        if len(df) == 0:
            print("❌ No data to analyze")
            return None
        
        print(f"\n✅ Collected {len(df)} data points from {processed} tickers")
        print(f"   Features extracted: {len([c for c in df.columns if c not in ['ticker', 'date', 'next_day_return', 'is_winner']])}")
        
        # DISCOVER PATTERNS
        print("\n🔍 DISCOVERING PATTERNS...\n")
        
        # Find which features correlate with winners
        feature_cols = [c for c in df.columns if c not in ['ticker', 'date', 'next_day_return', 'is_winner']]
        
        correlations = {}
        for feature in feature_cols:
            if df[feature].std() > 0:  # Only if feature varies
                corr = df[feature].corr(df['next_day_return'])
                if not pd.isna(corr):
                    correlations[feature] = corr
        
        # Sort by absolute correlation
        sorted_corr = sorted(correlations.items(), key=lambda x: abs(x[1]), reverse=True)
        
        print("🏆 TOP 15 PREDICTIVE FEATURES (Unknown patterns discovered!):")
        print(f"{'Feature':<30} {'Correlation':<12} {'Signal'}")
        print("-"*70)
        for feature, corr in sorted_corr[:15]:
            signal = "📈 Buy when HIGH" if corr > 0 else "📉 Buy when LOW"
            print(f"{feature:<30} {corr:>+.4f}       {signal}")
        
        # Find feature combinations (interactions)
        print("\n🧪 TESTING FEATURE COMBINATIONS...")
        self._discover_interactions(df, feature_cols[:10])  # Test top 10 features
        
        # Build winner profile
        print("\n🎯 WINNER PROFILE (Data-driven rules):")
        winners = df[df['is_winner'] == True]
        losers = df[df['is_winner'] == False]
        
        for feature in sorted_corr[:10]:
            feature_name = feature[0]
            winner_avg = winners[feature_name].mean()
            loser_avg = losers[feature_name].mean()
            diff = winner_avg - loser_avg
            
            if abs(diff) > 0.01:  # Meaningful difference
                if diff > 0:
                    print(f"   ✅ Winners have HIGHER {feature_name}: {winner_avg:.3f} vs {loser_avg:.3f}")
                else:
                    print(f"   ✅ Winners have LOWER {feature_name}: {winner_avg:.3f} vs {loser_avg:.3f}")
        
        self.feature_importance = dict(sorted_corr)
        return df
    
    def _discover_interactions(self, df, features):
        """Find feature COMBINATIONS that predict winners"""
        best_combo = None
        best_corr = 0
        
        # Try a few promising combinations
        for i in range(min(5, len(features))):
            for j in range(i+1, min(5, len(features))):
                feat1, feat2 = features[i], features[j]
                
                # Try multiplication
                combo = df[feat1] * df[feat2]
                corr = combo.corr(df['next_day_return'])
                
                if not pd.isna(corr) and abs(corr) > abs(best_corr):
                    best_corr = corr
                    best_combo = f"{feat1} × {feat2}"
        
        if best_combo:
            print(f"   🔥 BEST COMBO: {best_combo} → {best_corr:+.4f} correlation")
    
    def generate_scanner(self, df, min_win_rate=0.60):
        """Build a data-driven scanner from discovered patterns"""
        print("\n🚀 GENERATING DATA-DRIVEN SCANNER...")
        
        # Use top predictive features
        top_features = list(self.feature_importance.items())[:5]
        
        print(f"\nScanner based on top 5 discovered features:")
        for feature, corr in top_features:
            threshold = df[feature].quantile(0.7) if corr > 0 else df[feature].quantile(0.3)
            operator = ">" if corr > 0 else "<"
            print(f"   {feature} {operator} {threshold:.3f}")
        
        print(f"\n💡 USE THIS: These rules are DATA-DRIVEN, not opinions!")
        return top_features

discovery_engine = PatternDiscoveryEngine()
print("✅ Pattern Discovery Engine ready")
print("   This will find patterns WE DON'T KNOW about yet")
```

# 🎲 ANOMALY HUNTER: Find The Weird Stuff That Works

**Mission:** Find patterns that SHOULDN'T work but DO

**Method:**
- Look for counter-intuitive correlations
- Find tickers that behave opposite to their sector
- Discover "broken" patterns that actually win
- Test ridiculous ideas that the data supports

**Why:** The best edges are the ones that don't make sense - until they do.

**Examples to test:**
- Do some tickers win MORE when their sector loses?
- Does low volume sometimes predict big moves?
- Do down streaks in some tickers = buy signals?

**This is the UNKNOWN we're looking for.**

```python
class AnomalyHunter:
    """
    Find the WEIRD patterns that work
    
    Contrarian logic + data mining = hidden edges
    """
    
    def __init__(self):
        self.anomalies = []
    
    def find_sector_contrarians(self, tickers, days_back=60):
        """Find tickers that WIN when their SUB-SECTOR LOSES"""
        print("\n🎲 HUNTING: Sub-Sector Contrarians")
        print("="*70)
        print("Looking for tickers that move OPPOSITE to their sub-sector")
        print("(KDK vs trucking peers, not vs all AVs)\n")
        
        # Group tickers by SUB-SECTOR (fine granularity)
        sector_tickers = {}
        for ticker in tickers:
            sub_sector = get_sub_sector(ticker)
            if sub_sector not in sector_tickers:
                sector_tickers[sub_sector] = []
            sector_tickers[sub_sector].append(ticker)
        
        contrarians = []
        for sub_sector, sector_list in sector_tickers.items():
            if len(sector_list) < 2:
                continue
            
            print(f"   Testing {sub_sector} sub-sector ({len(sector_list)} tickers)...")
            
            # Get sector average performance
            sector_data = {}
            for ticker in sector_list:
                data = get_data(ticker, days_back=days_back)
                if data is not None and len(data) > 0:
                    sector_data[ticker] = data['Close'].pct_change()
            
            if len(sector_data) < 2:
                continue
            
            # Calculate SUB-SECTOR average returns
            sector_avg = pd.DataFrame(sector_data).mean(axis=1)
            
            # Find tickers with NEGATIVE correlation to SUB-SECTOR
            for ticker, returns in sector_data.items():
                # Align dates
                common_idx = sector_avg.index.intersection(returns.index)
                if len(common_idx) < 20:
                    continue
                
                corr = returns.loc[common_idx].corr(sector_avg.loc[common_idx])
                
                # Test: Does it win when SUB-SECTOR loses?
                sector_down = sector_avg.loc[common_idx] < -0.02
                ticker_returns_on_sector_down = returns.loc[common_idx][sector_down]
                
                if len(ticker_returns_on_sector_down) > 5:
                    win_rate = (ticker_returns_on_sector_down > 0).mean()
                    avg_return = ticker_returns_on_sector_down.mean()
                    
                    if win_rate > 0.55:
                        contrarians.append({
                            'ticker': ticker,
                            'sub_sector': sub_sector,
                            'correlation': corr,
                            'win_rate_when_sector_down': win_rate,
                            'avg_return_when_sector_down': avg_return,
                            'sample_size': len(ticker_returns_on_sector_down)
                        })
        
        if len(contrarians) > 0:
            df = pd.DataFrame(contrarians).sort_values('win_rate_when_sector_down', ascending=False)
            print(f"\n🔥 FOUND {len(df)} SUB-SECTOR CONTRARIANS!")
            print("\nThese tickers WIN when their SUB-SECTOR LOSES:\n")
            for _, row in df.iterrows():
                print(f"   {row['ticker']:6s} ({row['sub_sector']:25s}): {row['win_rate_when_sector_down']:.0%} WR, {row['avg_return_when_sector_down']:+.2f}% avg")
                print(f"          Sub-sector correlation: {row['correlation']:+.2f} (NEGATIVE = contrarian)")
            
            return df
        else:
            print("   No strong contrarians found")
            return None
    
    def find_silent_movers(self, tickers, days_back=60):
        """Find tickers that make BIG moves on LOW volume (institutional accumulation?)"""
        print("\n🎲 HUNTING: Silent Movers")
        print("="*70)
        print("Looking for big price moves on suspiciously low volume\n")
        
        silent_movers = []
        processed = 0
        
        for ticker in tickers:
            data = get_data(ticker, days_back=days_back)
            if data is None or len(data) < 30:
                continue
            
            processed += 1
            
            # Calculate volume MA
            vol_ma = data['Volume'].rolling(20).mean()
            price_change = data['Close'].pct_change()
            
            # Find days with big move (>3%) on low volume (<0.7x MA)
            big_moves_low_vol = ((abs(price_change) > 0.03) & (data['Volume'] < vol_ma * 0.7))
            
            if big_moves_low_vol.sum() > 2:
                # What happens NEXT day after silent move?
                next_day_returns = []
                for i in range(len(data)-1):
                    if big_moves_low_vol.iloc[i]:
                        next_day_ret = (data['Close'].iloc[i+1] / data['Close'].iloc[i]) - 1
                        next_day_returns.append(next_day_ret)
                
                if len(next_day_returns) > 2:
                    win_rate = sum([r > 0 for r in next_day_returns]) / len(next_day_returns)
                    avg_return = np.mean(next_day_returns)
                    
                    if win_rate > 0.60:  # Silent movers predict continuation
                        silent_movers.append({
                            'ticker': ticker,
                            'silent_moves': big_moves_low_vol.sum(),
                            'next_day_win_rate': win_rate,
                            'next_day_avg_return': avg_return,
                            'sample_size': len(next_day_returns)
                        })
        
        if len(silent_movers) > 0:
            df = pd.DataFrame(silent_movers).sort_values('next_day_win_rate', ascending=False)
            
            print(f"\n🔥 FOUND {len(df)} SILENT MOVERS!")
            print("\nThese tickers make big moves on LOW volume (then continue):\n")
            for _, row in df.head(10).iterrows():
                print(f"   {row['ticker']:6s}: {row['next_day_win_rate']:.0%} WR next day, {row['next_day_avg_return']:+.2f}% avg")
                print(f"          Silent moves detected: {row['silent_moves']} times")
            
            return df
        else:
            print(f"   No strong patterns found in {processed} tickers")
            return None
    
    def find_reverse_momentum(self, tickers, days_back=60):
        """Find tickers where DOWN streaks predict UP moves (mean reversion edge)"""
        print("\n🎲 HUNTING: Reverse Momentum")
        print("="*70)
        print("Looking for tickers where losing streaks = buying opportunity\n")
        
        reverse_momentum = []
        processed = 0
        
        for ticker in tickers:
            data = get_data(ticker, days_back=days_back)
            if data is None or len(data) < 30:
                continue
            
            processed += 1
            
            daily = data['Close'].pct_change()
            
            # Find 2-day losing streaks
            two_day_losses = []
            for i in range(2, len(daily)-1):
                if daily.iloc[i-1] < -0.01 and daily.iloc[i] < -0.01:  # 2 down days
                    next_day_return = daily.iloc[i+1]
                    two_day_losses.append(next_day_return)
            
            if len(two_day_losses) > 5:
                win_rate = sum([r > 0 for r in two_day_losses]) / len(two_day_losses)
                avg_return = np.mean(two_day_losses)
                
                if win_rate > 0.60 and avg_return > 0.015:  # Strong reversal
                    reverse_momentum.append({
                        'ticker': ticker,
                        'win_rate_after_2_down': win_rate,
                        'avg_return_after_2_down': avg_return,
                        'sample_size': len(two_day_losses)
                    })
        
        if len(reverse_momentum) > 0:
            df = pd.DataFrame(reverse_momentum).sort_values('win_rate_after_2_down', ascending=False)
            
            print(f"\n🔥 FOUND {len(df)} REVERSE MOMENTUM TICKERS!")
            print("\nThese tickers BOUNCE after 2 down days:\n")
            for _, row in df.head(10).iterrows():
                print(f"   {row['ticker']:6s}: {row['win_rate_after_2_down']:.0%} WR, {row['avg_return_after_2_down']:+.2f}% avg")
                print(f"          Tested on {row['sample_size']} 2-day losing streaks")
            
            return df
        else:
            print(f"   No strong patterns found in {processed} tickers")
            return None

anomaly_hunter = AnomalyHunter()
print("✅ Anomaly Hunter ready")
print("   This finds the WEIRD patterns that shouldn't work - but do")
```

# TEST 3: YOUR SCANNER 1 (VOLUME BREAKOUT)

**Your Scanner:** Volume >20x + Price >5%

**Test:** Forward validation with 1-day lag

```python
def test_scanner1_forward(tickers_to_test, days_back=30):
    """
    Test YOUR Scanner 1 with forward lag - FULL UNIVERSE
    Scanner triggers Day N → Enter Day N+1 → Measure Day N+2
    """
    print("Testing YOUR SCANNER 1: Volume >20x + Price >5%")
    print("="*70)
    print(f"Testing {len(tickers_to_test)} tickers...")
    
    all_signals = []
    processed = 0
    
    for ticker in tickers_to_test:
        data = get_data(ticker, days_back=days_back)
        if data is None:
            continue
        
        processed += 1
        if processed % 20 == 0:
            print(f"   Processed {processed}/{len(tickers_to_test)}...")
        
        daily = data.resample('1D').agg({
            'Open': 'first',
            'High': 'max',
            'Low': 'min',
            'Close': 'last',
            'Volume': 'sum'
        }).dropna()
        
        # Calculate volume MA
        daily['vol_ma'] = daily['Volume'].rolling(20).mean()
        daily['pct_change'] = daily['Close'].pct_change() * 100
        
        # Find Scanner 1 signals
        for i in range(20, len(daily)-2):
            signal_day = daily.iloc[i]
            vol_ratio = signal_day['Volume'] / signal_day['vol_ma']
            pct_change = signal_day['pct_change']
            
            # Scanner 1 criteria
            if vol_ratio > 5 and pct_change > 3:
                entry_day = daily.iloc[i+1]
                outcome_day = daily.iloc[i+2]
                
                # Forward return
                forward_return = ((outcome_day['Close'] - entry_day['Open']) / entry_day['Open']) * 100
                
                all_signals.append({
                    'ticker': ticker,
                    'date': daily.index[i],
                    'vol_ratio': vol_ratio,
                    'signal_day_gain': pct_change,
                    'forward_return': forward_return,
                    'win': forward_return > 0
                })
    
    df_signals = pd.DataFrame(all_signals)
    
    if len(df_signals) == 0:
        print(f"❌ No Scanner 1 signals found from {processed} tickers")
        return None
    
    print(f"\n📊 FOUND {len(df_signals)} Scanner 1 signals from {processed} tickers")
    print(f"\nForward Performance (1-day lag):")
    print(f"   Win Rate: {df_signals['win'].mean():.1%}")
    print(f"   Avg Return: {df_signals['forward_return'].mean():+.1f}%")
    print(f"   Median Return: {df_signals['forward_return'].median():+.1f}%")
    print(f"   Best: {df_signals['forward_return'].max():+.1f}%")
    print(f"   Worst: {df_signals['forward_return'].min():+.1f}%")
    
    # TOP PERFORMERS
    top_tickers = df_signals.groupby('ticker').agg({
        'win': 'mean',
        'forward_return': 'mean',
        'ticker': 'count'
    }).rename(columns={'ticker': 'signals'})
    top_tickers = top_tickers[top_tickers['signals'] >= 2]  # At least 2 signals
    top_tickers = top_tickers.sort_values('win', ascending=False).head(15)
    
    print(f"\n🏆 TOP 15 TICKERS for Scanner 1:")
    for ticker, row in top_tickers.iterrows():
        print(f"   {ticker:6s}: {row['win']:.0%} WR, {row['forward_return']:+.1f}% avg, {int(row['signals'])} signals")
    
    # Top 5 individual signals
    print(f"\nBest Individual Signals:")
    top5 = df_signals.nlargest(5, 'forward_return')
    for _, row in top5.iterrows():
        print(f"   {row['ticker']:6s} {row['date'].date()} → {row['forward_return']:+.1f}%")
    
    return df_signals

print("✅ Scanner 1 tester ready")
```

```python
def test_sector_follow_the_leader(sector_name, days_back=60):
    """YOUR INSIGHT: When leader spikes, followers catch up - SUB-SECTOR level"""
    print(f"\n🎲 TESTING: Follow-The-Leader ({sector_name} SUB-SECTOR)")
    print("="*70)
    
    tickers = SUB_SECTORS.get(sector_name, [])
    if len(tickers) < 2:
        print(f"❌ Need multiple tickers in sub-sector (only {len(tickers)} found)")
        return None
    
    # Get data for all
    sector_data = {}
    for ticker in tickers:
        data = get_data(ticker, days_back=days_back)
        if data is not None:
            daily = data.resample('1D').agg({
                'Open': 'first',
                'Close': 'last',
                'Volume': 'sum'
            }).dropna()
            daily['pct_change'] = daily['Close'].pct_change() * 100
            sector_data[ticker] = daily
    
    if len(sector_data) < 2:
        print("❌ Not enough sector data")
        return None
    
    signals = []
    
    # For each ticker, see if others follow when it spikes
    for leader in sector_data:
        leader_df = sector_data[leader]
        
        for i in range(1, len(leader_df)-2):
            leader_spike = leader_df['pct_change'].iloc[i]
            
            if leader_spike > 8:  # Leader spikes >8%
                # Check if followers move next day
                for follower in sector_data:
                    if follower == leader:
                        continue
                    
                    follower_df = sector_data[follower]
                    
                    # Align dates
                    if leader_df.index[i] in follower_df.index:
                        idx = follower_df.index.get_loc(leader_df.index[i])
                        
                        if idx < len(follower_df) - 2:
                            # Enter follower next day
                            entry = follower_df['Open'].iloc[idx+1]
                            exit_price = follower_df['Close'].iloc[idx+2]
                            forward_return = ((exit_price - entry) / entry) * 100
                            
                            signals.append({
                                'leader': leader,
                                'leader_spike': leader_spike,
                                'follower': follower,
                                'date': leader_df.index[i],
                                'forward_return': forward_return,
                                'win': forward_return > 0
                            })
    
    if len(signals) == 0:
        print("❌ No leader-follower signals")
        return None
    
    df = pd.DataFrame(signals)
    win_rate = df['win'].mean()
    avg_return = df['forward_return'].mean()
    
    print(f"\n📊 RESULTS: {len(df)} follower trades")
    print(f"   Win Rate: {win_rate:.1%}")
    print(f"   Avg Return: {avg_return:+.2f}%")
    print(f"\n   Top pairs:")
    pair_performance = df.groupby(['leader', 'follower']).agg({
        'win': 'mean',
        'forward_return': 'mean'
    }).sort_values('win', ascending=False).head(3)
    print(pair_performance)
    
    if win_rate > 0.60:
        verdict = "WINNER"
        lesson = f"Sub-sector momentum LAG is REAL in {sector_name} - trade the followers!"
    elif win_rate > 0.50:
        verdict = "MAYBE"
        lesson = "Followers sometimes catch up - refine which ones"
    else:
        verdict = "LOSER"
        lesson = "Sub-sector momentum doesn't lag - leaders stay leaders"
    
    print(f"\n🎯 VERDICT: {verdict}")
    print(f"   Lesson: {lesson}")
    
    save_test_result(
        name=f"Follow The Leader ({sector_name})",
        approach="Hybrid (Your Insight)",
        win_rate=win_rate,
        avg_return=avg_return,
        sample_size=len(df),
        verdict=verdict,
        lesson=lesson
    )
    
    return df

# Test on quantum_hardware (IONQ, RGTI, QMCO)
quantum_leader_results = test_sector_follow_the_leader('quantum_hardware', days_back=60)
```

```python
def test_sector_momentum(sector_name, days_back=60):
    """
    Test if SUB-SECTOR stocks move together.
    YOUR INSIGHT: KDK (trucking) ≠ robotaxis - need sub-sector precision!
    When leader spikes, do followers spike within 1-2 days?
    """
    print(f"\n🎲 TESTING: SUB-SECTOR MOMENTUM: {sector_name}")
    print("="*70)
    
    tickers = SUB_SECTORS.get(sector_name, [])
    if len(tickers) < 2:
        print(f"❌ Need at least 2 tickers in sub-sector (only {len(tickers)} found)")
        return None
    
    # Get data for all tickers
    sector_data = {}
    for ticker in tickers:
        data = get_data(ticker, days_back=days_back)
        if data is not None:
            daily = data.resample('1D').agg({
                'Close': 'last',
                'Volume': 'sum'
            }).dropna()
            daily['pct_change'] = daily['Close'].pct_change() * 100
            sector_data[ticker] = daily
    
    if len(sector_data) < 2:
        print("❌ Not enough data")
        return None
    
    print(f"\n📊 Loaded {len(sector_data)} tickers from sub-sector '{sector_name}'")
    
    # Find correlation between moves
    correlations = []
    
    for ticker1 in sector_data:
        for ticker2 in sector_data:
            if ticker1 >= ticker2:  # Avoid duplicates
                continue
            
            df1 = sector_data[ticker1]
            df2 = sector_data[ticker2]
            
            # Align dates
            common_dates = df1.index.intersection(df2.index)
            if len(common_dates) < 10:
                continue
            
            pct1 = df1.loc[common_dates, 'pct_change']
            pct2 = df2.loc[common_dates, 'pct_change']
            
            # Correlation
            corr = pct1.corr(pct2)
            
            # Count times they moved together (both >5% same day)
            both_up = ((pct1 > 5) & (pct2 > 5)).sum()
            
            correlations.append({
                'pair': f"{ticker1}+{ticker2}",
                'correlation': corr,
                'both_up_days': both_up,
                'total_days': len(common_dates)
            })
    
    df_corr = pd.DataFrame(correlations)
    df_corr = df_corr.sort_values('correlation', ascending=False)
    
    print(f"\nSub-Sector Correlations ('{sector_name}'):")
    for _, row in df_corr.head(5).iterrows():
        print(f"   {row['pair']:15s} | Corr: {row['correlation']:.2f} | Both up: {row['both_up_days']} days")
    
    avg_corr = df_corr['correlation'].mean()
    print(f"\n🎯 Average correlation: {avg_corr:.2f}")
    
    if avg_corr > 0.5:
        print(f"   ✅ STRONG sub-sector momentum - {sector_name} stocks move together!")
        lesson = f"Sub-sector '{sector_name}' has tight correlation - trade as group"
    elif avg_corr > 0.3:
        print(f"   ⚠️  MODERATE sub-sector momentum in {sector_name}")
        lesson = f"Some correlation in '{sector_name}' - selective pairs work"
    else:
        print(f"   ❌ WEAK sub-sector momentum - {sector_name} stocks move independently")
        lesson = f"Sub-sector '{sector_name}' doesn't move together - trade individually"
    
    return df_corr

print("✅ Sub-sector momentum tester ready (updated for precision)")
```
